```python
graph = {
    'B' : ['A', 'F'],
    'A' : [],
    'F' : ['D', 'E', 'H'],
    'D' : [],
    'E' : ['U', 'S'],
    'H' : [],
    'U' : [],
    'S' : ['Z', 'W'],
    'Z' : [],
    'W' : []
}

from collections import deque

def bfs(graph, start):

  v = []            #stores visited nodes
  queue = deque()         #creates empty queue

  v.append(start)          #visit start node
  queue.append(start)            #add start node to queue

  while queue:

    node = queue.popleft()     #removes leftmost node
    print(node, end=" ")

    #visit all neighbours
    for n in graph[node]:
      if n not in v:
        v.append(n)
        queue.append(n)

start = input("Enter the start node:").upper()

print("BFS Traversal:")
bfs(graph, start)
```

```python
graph = {
    'B' : ['A', 'F'],
    'A' : [],
    'F' : ['D', 'E', 'H'],
    'D' : [],
    'E' : ['U', 'S'],
    'H' : [],
    'U' : [],
    'S' : ['Z', 'W'],
    'Z' : [],
    'W' : []
}

def dfs(graph, start):
  v = []
  stack = []

  v.append(start)
  stack.append(start)

  while stack:
    node = stack.pop()
    print(node, end=" ")

    for n in graph[node]:
      if n not in v:
        v.append(n)
        stack.append(n)

start = input("Enter the start node:").upper()

print("DFS Traversal:")
dfs(graph, start)
```
