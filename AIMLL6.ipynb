```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)
```

```python
# Load Dataset
df = sns.load_dataset('titanic')
print("Initial Shape:", df.shape)
print("Missing Values Before Preprocessing:\n", df.isnull().sum())
```

```python
# Handling Missing Values
df = df.drop(columns=['deck', 'embark_town', 'alive'])

df['age'] = df['age'].fillna(df['age'].median())

df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])

print("\nMissing Values After Imputation:\n", df.isnull().sum())
```

```python
# Categorical Encoding
# Label Encoding
le = LabelEncoder()
df['sex'] = le.fit_transform(df['sex'])
df['alone'] = le.fit_transform(df['alone'])

# One-Hot Encoding
df = pd.get_dummies(df, columns=['embarked', 'class', 'who'], drop_first=True)
```

```python
# Normalization/Feature Scaling
scaler = MinMaxScaler()
df[['age', 'fare']] = scaler.fit_transform(df[['age', 'fare']])

print("\nProcessed Dataset Preview:")
print(df[['survived', 'pclass', 'sex', 'age', 'fare']].head())
```
